# Specifying date range, services and save directory

```python
from datetime import datetime
import os
from zoneinfo import ZoneInfo


%matplotlib inline

# Define tz-aware start and end dates using zoneinfo
start_date = datetime(2021, 2, 1)  # Start date with UTC timezone
end_date = datetime(2024, 12, 1)  # End date with UTC timezone


services = ['OpenAI:API', 'OpenAI:ChatGPT', 'OpenAI:DALL-E', 'OpenAI:Playground', 'Anthropic:API', 'Anthropic:Claude', 'Anthropic:Console', 'Character.AI:Character.AI', 'StabilityAI:REST', 'StabilityAI:gRPC', 'StabilityAI:Assistant']


# Get the project root by finding 'distributed-systems-lab' in the current path
current_path = os.getcwd()
project_root = current_path.split('distributed-systems-lab')[0] + 'distributed-systems-lab'

# Define figures directory
figures_dir = os.path.join(project_root, 'figures')

# Create directory if it doesn't exist
os.makedirs(figures_dir, exist_ok=True)
```

# MTBF and MTTR

```python
from llm_analysis.server.scripts.analysis_modules.mtbf_provider import analyze_mtbf_provider
from llm_analysis.server.scripts.analysis_modules.mttr_provider import analyze_mttr_provider

fig_mtbf = analyze_mtbf_provider(start_date, end_date, services)
fig_mttr = analyze_mttr_provider(start_date, end_date, services)

# Save the figures
fig_mtbf.savefig(os.path.join(figures_dir, 'mtbf_provider.pdf'), format='pdf', bbox_inches='tight')
fig_mttr.savefig(os.path.join(figures_dir, 'mttr_provider.pdf'), format='pdf', bbox_inches='tight')

# Verify files were saved
print(f"Files saved in: {figures_dir}")
print("Saved files:", os.listdir(figures_dir))
```

# Incident Co-occurrences

```python
from llm_analysis.server.scripts.analysis_modules.service_incidents import analyze_service_incidents

# Define tz-aware start and end dates using zoneinfo
start_date = datetime(2020, 2, 10)  # Start date with UTC timezone
end_date = datetime(2025, 1, 28)  # End date with UTC timezone

dict_services_incidents = analyze_service_incidents(start_date, end_date, services)
for key, value in dict_services_incidents.items():
    value.savefig(os.path.join(figures_dir, f'{key}_service_incidents.pdf'), format='pdf', bbox_inches='tight')
```
